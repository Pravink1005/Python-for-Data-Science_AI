## Return

Instead of just printing data, functions can send results back to the caller using the `return` keyword. Once a `return` statement is executed, the function exits immediately.

```python
def add(a,b):
    return a+b
print(add(2,3))
```

## Default Arguments

You can assign default values to parameters. If the caller doesn't provide an argument for that parameter, the default value is used automatically.

```python
def power(x,p=2):
    return x**p
print(power(5),power(5,3))
```

## Keyword Arguments

When calling a function, you can pass arguments using the name of the parameter (`key=value`). This means order doesn't matter.

```python
def info(name,age):
    print(name,age)
info(age=20,name='Sam')
```

## *args

Using `*args` allows a function to accept any number of positional arguments. Python packs these extra arguments into a tuple.

```python
def total(*nums):
    print(sum(nums))
total(1,2,3)
```

## **kwargs

Using `**kwargs` allows a function to accept any number of keyword (named) arguments. Python packs these into a dictionary.

```python
def show(**d):
    print(d)
show(name='A',city='Madurai')
```

## Scope

**Core Concept**: Scope determines the visibility of variables. 
  * **Local variables** are created inside a function and cannot be accessed outside.
  * **Global variables** are created in the main body and can be referenced globally.

```python
x=10
def f():
    x=5
    print(x)
f();print(x)
```

## Recursion

Recursion happens when a function calls itself to solve a smaller instance of the same problem. Every recursive function must have a base case to stop infinite loops.

```python
def fact(n):
    return 1 if n<=1 else n*fact(n-1)
print(fact(5))
```

Build a mini-project that combines functions with conditionals and loops from your basics curriculum

```python
def calculate_grade(marks):
  if marks >= 90:
    return "A"
  elif marks >= 75:
    return "B"
  elif marks >= 50:
    return "C"
  else:
    return "Fail"


def basic_calculator(a, b, op):
  if op == "+":
    return a + b
  elif op == "-":
    return a - b
  elif op == "*":
    return a * b
  elif op == "/":
    return a / b if b != 0 else "Error: Division by zero"
  else:
    return "Invalid Operation"


# Main execution loop
while True:
  choice = input(
      "Choose option (1: Grade, 2: Calculator, q: Quit): "
  )
  if choice == "q":
    print("Exiting program. Done.")
    break
  elif choice == "1":
    m = float(input("Enter marks: "))
    print(f"Grade: {calculate_grade(m)}")
  elif choice == "2":
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    operator = input("Enter operator (+, -, *, /): ")
    print(f"Result: {basic_calculator(num1, num2, operator)}")
```

Build a mini-project that combines variables, conditional statements, and functions to manage a simple student directory.

```python
def check_status(score):
  if score >= 50:
    return "Passed"
  else:
    return "Failed"


def add_student(name, score, **details):
  status = check_status(score)
  print(f"\n--- Student Record: {name} ---")
  print(f"Score: {score} ({status})")
  for key, value in details.items():
    print(f"{key.capitalize()}: {value}")


# Main execution loop
while True:
  action = input("\nDo you want to add a student? (yes/no): ")
  if action.lower() == "no":
    print("Exiting directory. Done.")
    break
  elif action.lower() == "yes":
    student_name = input("Enter student name: ")
    student_score = float(input("Enter score: "))
    student_city = input("Enter city: ")

    add_student(
        student_name, student_score, city=student_city, status="Active"
    )
  else:
    print("Invalid input, please type yes or no.")
```

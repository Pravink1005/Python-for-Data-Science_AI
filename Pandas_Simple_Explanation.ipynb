## 11. Selecting Rows with `loc`

`loc` is used for **label-based selection**.

Example:

```python
df.loc[0]
```

gets row with index 0.

You can also select particular rows and columns:

```python
df.loc[0:2, ["Name", "Mark"]]
```

````python
import pandas as pd

# Sample DataFrame
data = {'name': ['Alice', 'Bob', 'Charlie'], 'age': [25, 30, 35]}
df = pd.DataFrame(data, index=['row1', 'row2', 'row3'])

# Select a single row by label
print(df.loc['row1'])

# Select specific rows and columns by label
print(df.loc['row1':'row3', ['name', 'age']])  # 'row3' is included (inclusive)
````

## 12. Selecting Rows with `iloc`

`iloc` uses **position numbers**.

Example:

```python
df.iloc[0]
```

means the first row.

```python
df.iloc[0:2]
```

means the first two rows.

````python
# Select the first row (index position 0)
print(df.iloc[0])

# Select rows from position 0 up to 2 (excludes position 2) and columns 0 to 1
print(df.iloc[0:2, 0:2])
````

### `loc` vs `iloc`

Easy memory trick:

```text
loc  → label
iloc → integer position
```

## 17. Creating a Column from a Calculation

We can calculate a new column from existing columns.

Example:

```python

```

For this simple dataset, the mark is already out of 100, so the value is the same.

In real datasets, we might calculate:
- total
- average
- profit
- age
- percentage
- tax

from existing columns.

````python
import pandas as pd

# Sample DataFrame
data = {'product': ['A', 'B', 'C'],
        'price': [100, 200, 300],
        'quantity': [2, 5, 3]}
df = pd.DataFrame(data)

# Calculate a new column (Price * Quantity)
df['total_cost'] = df['price'] * df['quantity']

print(df)
````

## 19. Missing Values

Real-world datasets often contain missing values.

Pandas represents missing values using `NaN` in many situations.

Example dataset:

```text
Name    Age    Mark
Arun    20     80
Bala    NaN    75
Cathy   21     NaN
David   22     65
```

````python
student_data = pd.DataFrame({
    "Name": ["Arun", "Bala", "Cathy", "David"],
    "Age": [20, None, 21, 22],
    "Mark": [80, 75, None, 65]
})

print(student_data)
````

## 20. Finding Missing Values

### `isna()`

```python
df.isna()
```

shows `True` where a value is missing.

### Count missing values

```python
df.isna().sum()
```

This tells us how many missing values each column contains.

````python
print("Missing value locations:")
print(student_data.isna())

print("\nMissing values in each column:")
print(student_data.isna().sum())
````

## 21. Removing Missing Rows

`dropna()` removes rows containing missing values.

```python
clean_df = df.dropna()
```

Use this when removing incomplete records makes sense.

Be careful: you may lose useful data.

````python
without_missing = student_data.dropna()

print(without_missing)
````

## 22. Filling Missing Values

Instead of removing rows, we can fill missing values.

Example:

```python
df["Age"].fillna(df["Age"].mean())
```

This replaces missing ages with the average age.

For marks, we can use the median:

```python
df["Mark"].fillna(df["Mark"].median())
```

````python
filled = student_data.copy()

filled["Age"] = filled["Age"].fillna(filled["Age"].mean())
filled["Mark"] = filled["Mark"].fillna(filled["Mark"].median())

print(filled)
````

## 23. Duplicate Rows

Real datasets can contain duplicate records.

### Find duplicates

```python
df.duplicated()
```

### Count duplicates

```python
df.duplicated().sum()
```

### Remove duplicates

```python
df.drop_duplicates()
```

````python
duplicate_data = pd.DataFrame({
    "Name": ["Arun", "Bala", "Bala", "Cathy"],
    "Mark": [80, 75, 75, 90]
})

print("Duplicate rows:")
print(duplicate_data[duplicate_data.duplicated()])

print("\nNumber of duplicates:", duplicate_data.duplicated().sum())

print("\nAfter removing duplicates:")
print(duplicate_data.drop_duplicates())
````

## 25. GroupBy

`groupby()` is one of the most important Pandas concepts.

Suppose we have:

```text
Department   Salary
IT           40000
IT           50000
HR           35000
HR           45000
```

We can group by department and calculate the average salary.

```python
df.groupby("Department")["Salary"].mean()
```

### Easy idea

> **GroupBy = Group similar rows, then calculate something.**

````python
employees = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Salary": [40000, 50000, 35000, 45000, 30000, 42000]
})

average_salary = employees.groupby("Department")["Salary"].mean()

print(average_salary)
````

## 26. GroupBy with Multiple Statistics

We can calculate more than one statistic:

```python
grouped = employees.groupby("Department")["Salary"].agg(
    ["count", "mean", "min", "max"]
)
```

This gives a useful summary for each group.

````python
summary = employees.groupby("Department")["Salary"].agg(
    ["count", "mean", "min", "max"]
)

print(summary)
````

## 27. Aggregation

Common aggregation functions:

```text
sum()   → total
mean()  → average
median()→ middle value
min()   → smallest
max()   → largest
count() → number of values
std()   → standard deviation
```

````python
salary = employees["Salary"]

print("Total:", salary.sum())
print("Average:", salary.mean())
print("Median:", salary.median())
print("Minimum:", salary.min())
print("Maximum:", salary.max())
print("Count:", salary.count())
print("Standard deviation:", salary.std())
````

## 30. Merging DataFrames

Sometimes data is stored in separate tables.

Example:

### Student table

```text
Student_ID   Name
1            Arun
2            Bala
3            Cathy
```

### Marks table

```text
Student_ID   Mark
1            80
2            75
3            90
```

We can combine them using `merge()`.

```python
pd.merge(student_df, marks_df, on="Student_ID")
```

The common column is called the **key**.

````python
student_df = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Name": ["Arun", "Bala", "Cathy"]
})

marks_df = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Mark": [80, 75, 90]
})

merged = pd.merge(student_df, marks_df, on="Student_ID")

print(merged)
````

## 31. Concatenating DataFrames

`concat()` is useful when we want to put DataFrames together.

For example, two datasets containing students from different classes can be combined row by row.

````python
class_a = pd.DataFrame({
    "Name": ["Arun", "Bala"],
    "Mark": [80, 75]
})

class_b = pd.DataFrame({
    "Name": ["Cathy", "David"],
    "Mark": [90, 85]
})

all_students = pd.concat([class_a, class_b], ignore_index=True)

print(all_students)
````

## 33. Changing Data Types

Use `astype()` when we need to change a column's data type.

Example:

```python
df["Age"] = df["Age"].astype(int)
```

Be careful: conversion only works when the values can actually be converted.

````python
type_data = pd.DataFrame({
    "Age": ["20", "21", "22"]
})

print("Before:")
print(type_data.dtypes)

type_data["Age"] = type_data["Age"].astype(int)

print("\nAfter:")
print(type_data.dtypes)
````

# 35. Mini Practical Data Analysis

Now let's combine the important concepts.

We will:
1. Create a student dataset
2. Inspect the data
3. Calculate statistics
4. Filter high scorers
5. Group by department
6. Find the average score
7. Sort the results

````python
students = pd.DataFrame({
    "Name": ["Arun", "Bala", "Cathy", "David", "Esha", "Farhan"],
    "Department": ["IT", "IT", "HR", "HR", "IT", "Sales"],
    "Age": [20, 21, 22, 21, 20, 23],
    "Score": [85, 72, 90, 65, 95, 78]
})

print(students)
````

````python
department_average = students.groupby("Department")["Score"].mean()

print("Average score by department:")
print(department_average)
````

````python
sorted_students = students.sort_values("Score", ascending=False)

print("Students sorted by score:")
print(sorted_students)
````

````python
# Practice Answers

practice = pd.DataFrame({
    "Name": ["Arun", "Bala", "Cathy"],
    "Age": [20, 21, 22],
    "Score": [80, 75, 90],
    "Department": ["IT", "HR", "IT"]
})

print("DataFrame:")
print(practice)

print("\nFirst 2 rows:")
print(practice.head(2))

print("\nName and Score:")
print(practice[["Name", "Score"]])

print("\nScores greater than 80:")
print(practice[practice["Score"] > 80])

print("\nSorted by score:")
print(practice.sort_values("Score", ascending=False))

print("\nAverage score:", practice["Score"].mean())

practice["Pass"] = practice["Score"] >= 40
print("\nWith Pass column:")
print(practice)

print("\nAverage score by department:")
print(practice.groupby("Department")["Score"].mean())

practice.to_csv("pandas_practice_output.csv", index=False)
print("\nSaved: pandas_practice_output.csv")
````

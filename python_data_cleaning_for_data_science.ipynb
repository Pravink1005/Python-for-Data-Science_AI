## 1. What is Data Cleaning?

**Data cleaning** means finding and fixing problems in data before using it for analysis or machine learning.

Common problems:
- Missing values
- Duplicate rows
- Wrong spelling or inconsistent values
- Wrong data types
- Unusual values (outliers)

**Simple example:** If age is `-25`, something is wrong. We need to fix or remove it.

## 2. Import Pandas

````python
import pandas as pd

print("Pandas is ready!")
````

## 3. Create a Small Dataset

We intentionally create some problems so that we can practice cleaning them.

````python
data = {
    "Name": ["Arun", "Priya", "Arun", "Kumar", "Divya", "Ravi"],
    "Age": [22, 25, 22, None, 28, -5],
    "City": ["Chennai", "chennai", "Chennai", "Coimbatore", None, "Madurai"],
    "Salary": [25000, 30000, 25000, 35000, None, 40000]
}

df = pd.DataFrame(data)
df
````

## 4. Understand the Data

Before cleaning, we should look at the data.

````python
print("First 5 rows:")
display(df.head())

print("Shape:", df.shape)

print("\nData types:")
print(df.dtypes)
````

## 5. Find Missing Values

`isnull()` checks whether a value is missing.

`sum()` counts the missing values in each column.

````python
df.isnull().sum()
````

### Easy idea

If the result is:

```text
Age       1
City      1
Salary    1
```

it means those columns contain missing values.

## 6. Handle Missing Values

There are different ways to handle missing values.

- **Mean** → average value
- **Median** → middle value
- **Mode** → most common value
- **Drop** → remove the row/column

For salary, we can use the **median**.

````python
salary_median = df["Salary"].median()
df["Salary"] = df["Salary"].fillna(salary_median)

print("Median salary:", salary_median)
df
````

### Fill a missing text value

For `City`, we can use the most common city (**mode**).

````python
city_mode = df["City"].mode()[0]
df["City"] = df["City"].fillna(city_mode)

print("Most common city:", city_mode)
df
````

## 7. Remove Duplicate Rows

`duplicated()` finds duplicate rows.

`drop_duplicates()` removes them.

````python
print("Duplicate rows:")
print(df.duplicated())

df = df.drop_duplicates()

print("\nAfter removing duplicates:")
df
````

## 8. Fix Inconsistent Text Values

`Chennai` and `chennai` mean the same city, but Python sees them as different strings.

We can standardize the text using `str.title()`.

````python
df["City"] = df["City"].str.title()
df
````

## 9. Detect an Incorrect Value

Our dataset has an age of `-5`.

A person's age cannot normally be negative, so this is a bad value.

````python
print(df[df["Age"] < 0])
````

### Replace the incorrect age

Here we replace the negative age with the median age.

````python
valid_ages = df.loc[df["Age"] >= 0, "Age"]
age_median = valid_ages.median()

df.loc[df["Age"] < 0, "Age"] = age_median

df
````

## 10. Convert Data Types

Sometimes numbers are stored as text. We can convert them using `pd.to_numeric()`.

````python
df["Age"] = pd.to_numeric(df["Age"])
df["Salary"] = pd.to_numeric(df["Salary"])

print(df.dtypes)
````

## 11. Clean Column Names

Clean column names make code easier to write.

For example, `Employee Name` can become `employee_name`.

````python
df.columns = df.columns.str.lower().str.replace(" ", "_")

print(df.columns.tolist())
df
````

## 12. Detect Outliers — Simple Example

An **outlier** is a value that is very different from most other values.

Example: salaries `[25000, 27000, 30000, 28000, 500000]` → `500000` may be an outlier.

One common method is the **IQR (Interquartile Range)**.

````python
salary_data = pd.Series([25000, 27000, 30000, 28000, 500000])

Q1 = salary_data.quantile(0.25)
Q3 = salary_data.quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = salary_data[(salary_data < lower_limit) | (salary_data > upper_limit)]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Outliers:")
print(outliers)
````

## 13. Check the Clean Data

Always check the dataset again after cleaning.

````python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nFinal data:")
display(df)
````

## 14. Data Cleaning Cheat Sheet

| Problem | Pandas solution |
|---|---|
| Find missing values | `df.isnull().sum()` |
| Fill missing value | `df.fillna()` |
| Remove rows | `df.dropna()` |
| Find duplicates | `df.duplicated()` |
| Remove duplicates | `df.drop_duplicates()` |
| Change text | `df["column"].str.title()` |
| Convert number | `pd.to_numeric()` |
| Check data types | `df.dtypes` |
| Rename columns | `df.rename()` |
| Find outliers | IQR / Z-score |

### Remember
**Raw Data → Find Problems → Fix Problems → Check Again → Clean Data → Analysis/ML**

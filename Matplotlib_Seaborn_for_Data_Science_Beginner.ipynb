# Matplotlib & Seaborn for Data Science
### Beginner-friendly examples and practice

This notebook introduces the most useful plotting concepts for data science using:
- **Matplotlib** – basic and highly customizable plotting
- **Seaborn** – simple, attractive statistical visualizations

We will use small, easy-to-understand datasets and explain each example step by step.

````python
# Install if needed:
# !pip install matplotlib seaborn pandas numpy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Make plots appear inside the notebook
%matplotlib inline

print("Libraries imported successfully!")
````

## 1. Create a Simple Dataset

We will create a small student dataset and use it throughout the notebook.

````python
data = {
    "Student": ["Asha", "Bala", "Cathy", "David", "Esha", "Farhan"],
    "Study_Hours": [2, 4, 3, 5, 6, 4],
    "Marks": [55, 68, 62, 78, 88, 72],
    "Attendance": [75, 82, 80, 90, 95, 85],
    "Gender": ["F", "M", "F", "M", "F", "M"]
}

df = pd.DataFrame(data)
df
````

## 2. Matplotlib – Line Plot

A **line plot** is useful for showing trends or changes over an ordered variable.

````python
plt.plot(df["Student"], df["Marks"], marker="o")

plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.show()
````

### Understanding the code
- `plt.plot()` → creates a line chart
- `marker="o"` → shows a point for each value
- `plt.title()` → adds the chart title
- `plt.xlabel()` → names the X-axis
- `plt.ylabel()` → names the Y-axis
- `plt.show()` → displays the chart

## 3. Matplotlib – Bar Chart

A **bar chart** is useful for comparing categories.

````python
plt.bar(df["Student"], df["Marks"])

plt.title("Marks by Student")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.show()
````

## 4. Matplotlib – Scatter Plot

A **scatter plot** helps us understand the relationship between two numerical variables.

Here we will check whether more study hours are associated with higher marks.

````python
plt.scatter(df["Study_Hours"], df["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
````

### Simple interpretation

Look at the points from left to right.

If marks generally increase as study hours increase, there may be a **positive relationship** between study hours and marks.

## 5. Matplotlib – Histogram

A **histogram** shows the distribution of numerical values.

````python
plt.hist(df["Marks"], bins=4, edgecolor="black")

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
````

## 6. Matplotlib – Box Plot

A **box plot** helps identify the spread of data and possible outliers.

````python
plt.boxplot(df["Marks"])

plt.title("Marks Distribution")
plt.ylabel("Marks")

plt.show()
````

## 7. Adding a Grid and Figure Size

You can make a Matplotlib chart easier to read using `figsize()` and `grid()`.

````python
plt.figure(figsize=(8, 4))

plt.bar(df["Student"], df["Marks"])
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.grid(axis="y", alpha=0.3)

plt.show()
````

# Seaborn

Seaborn is built on top of Matplotlib and is especially useful for statistical data visualization.

A common workflow is:

```python
sns.chart_type(data=df, x="column", y="column")
plt.show()
```

## 8. Seaborn – Bar Plot

````python
sns.barplot(data=df, x="Student", y="Marks")

plt.title("Student Marks")
plt.show()
````

## 9. Seaborn – Scatter Plot

Seaborn makes it easy to add a third variable using `hue`.

Here, gender is used to distinguish the points.

````python
sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    hue="Gender"
)

plt.title("Study Hours vs Marks")
plt.show()
````

## 10. Seaborn – Histogram / Distribution Plot

`histplot()` shows the distribution of a numerical column.

````python
sns.histplot(data=df, x="Marks", bins=4)

plt.title("Distribution of Marks")
plt.show()
````

## 11. Seaborn – Box Plot

````python
sns.boxplot(data=df, y="Marks")

plt.title("Marks Distribution")
plt.show()
````

## 12. Seaborn – Count Plot

A **count plot** counts how many observations belong to each category.

````python
sns.countplot(data=df, x="Gender")

plt.title("Number of Students by Gender")
plt.show()
````

## 13. Seaborn – Relationship Between Attendance and Marks

Let's visualize another possible relationship.

````python
sns.scatterplot(
    data=df,
    x="Attendance",
    y="Marks",
    hue="Gender",
    s=100
)

plt.title("Attendance vs Marks")
plt.show()
````

## 14. Seaborn – Correlation Heatmap

A **correlation matrix** helps us see how numerical variables are related.

- Close to `+1` → strong positive relationship
- Close to `-1` → strong negative relationship
- Close to `0` → weak or no linear relationship

````python
numeric_df = df[["Study_Hours", "Marks", "Attendance"]]

correlation = numeric_df.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
````

## 15. Matplotlib vs Seaborn

| Feature | Matplotlib | Seaborn |
|---|---|---|
| Basic charts | Excellent | Excellent |
| Customization | Very high | High |
| Statistical plots | Possible | Very convenient |
| Beginner friendly | Moderate | Easy |
| Works with Pandas | Yes | Yes |
| Built on | Core plotting library | Matplotlib |

### Simple rule

**Matplotlib:** use when you need detailed control over a chart.

**Seaborn:** use when you want statistical visualizations quickly and cleanly.

# 16. Mini Data Science Example

### Question:
Does study time appear to be related to marks?

We can use a scatter plot and a trend line.

````python
sns.regplot(data=df, x="Study_Hours", y="Marks")

plt.title("Study Hours vs Marks with Trend Line")
plt.show()
````

### What to look for

The trend line gives us a simple visual indication of the relationship.

**Important:** A chart can show a relationship, but it does not by itself prove that studying more causes higher marks.

# 18. Quick Cheat Sheet

### Matplotlib
```python
plt.plot(x, y)                 # Line
plt.bar(x, y)                  # Bar
plt.scatter(x, y)              # Scatter
plt.hist(x)                    # Histogram
plt.boxplot(x)                 # Box plot
plt.title("Title")             # Title
plt.xlabel("X")                # X label
plt.ylabel("Y")                # Y label
plt.show()                     # Display
```

### Seaborn
```python
sns.lineplot(data=df, x="x", y="y")
sns.barplot(data=df, x="x", y="y")
sns.scatterplot(data=df, x="x", y="y")
sns.histplot(data=df, x="x")
sns.boxplot(data=df, x="x")
sns.countplot(data=df, x="x")
sns.heatmap(df.corr(), annot=True)
sns.regplot(data=df, x="x", y="y")
```

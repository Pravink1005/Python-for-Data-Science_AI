### What is Machine Learning?
It's type of artificial intelligence that teaches computers to learn from data and make decisions on their own instead of following hard-coded, step-by-step rules

### The Example Setup
Suppose we have the following data:

| Study Hours | Marks |
| :--- | :--- |
| 1 | 35 |
| 2 | 42 |
| 3 | 50 |
| 4 | 61 |
| 5 | 70 |
| 6 | 78 |
| 7 | 85 |

### 1. The Machine Learning Goal
We want the computer to learn the hidden pattern:
> *"If study hours increase, marks generally increase."*

### 2. The Prediction
Once the system learns this rule, we can feed it a new input:
* **Input:** `Study Hours = 8`
* **Predicted Output:** `Marks ≈ 92`


### Traditional Programming
* **Process:** Input + Rules $\rightarrow$ Output
* **How it works:** You give the computer data and write the specific instructions (code), and it generates the result.

### Machine Learning
* **Process:** Input + Output $\rightarrow$ Learn Rules
* **How it works:** You give the computer data and the correct answers, and it calculates the rules for you.


### Supervised Learning
The model learns from labeled data, meaning the relationship between input and output is already known.
* **Formula:** Input $\rightarrow$ Output
* **Example:** Study Hours $\rightarrow$ Marks
* **Common Algorithms:** `Linear Regression`, `Logistic Regression`, `Decision Tree`, `SVM`

***


# Supervised Learning Basics

### Why is it called "Supervised"?
> Because the model learns from known answers, just like a teacher supervising a student.

---

### How Supervised Learning Works
```text
Input Data + Correct Answer
          ↓
     ML Algorithm
          ↓
     Learn Pattern
          ↓
   New Input Data
          ↓
       Prediction
```

---

### Two Main Types of Supervised Learning

#### 1. Regression → *Predict a number*
* **Examples:**
  * Study Hours $\rightarrow$ Marks
  * House Size $\rightarrow$ House Price
  * Experience $\rightarrow$ Salary
* **Output Example:** `85 marks`

#### 2. Classification → *Predict a category*
* **Examples:**
  * Study Hours $\rightarrow$ Pass / Fail
  * Email $\rightarrow$ Spam / Not Spam
  * Customer $\rightarrow$ Buy / Not Buy
* **Output Example:** `Pass`

---

### 💡 Easy Definition for Students
**Supervised Learning** is a type of Machine Learning where a model learns from data that already contains the correct answers and uses what it learned to make predictions on new data.

> 📌 **Remember:**  
> `Supervised` = Data + Answer $\rightarrow$ Learn $\rightarrow$ Predict.

## 1. Linear Regression

### What does it do?
**Linear Regression predicts a number.**

* **Example:** Based on study hours, predict marks.

---

### The Data

| Study Hours | Marks |
| :---: | :---: |
| 1 | 35 |
| 2 | 45 |
| 3 | 55 |
| 4 | 65 |
| 5 | 75 |

---

### The Program

```python
from sklearn.linear_model import LinearRegression

# Data
X = [[1], [2], [3], [4], [5]]  #Why [[1], [2], [3]] instead of [1, 2, 3]? Because Scikit-learn expects the input in 2D format: Rows = students/data records Columns = features
y = [35, 45, 55, 65, 75]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)  #fit() means: Learn from the given data.

# Predict marks for 6 study hours
prediction = model.predict([[6]])

print("Predicted Marks:", prediction[0])
```

---

### Simple Explanation
$$\text{Study Hours} \longrightarrow \text{Linear Regression} \longrightarrow \text{Marks}$$

The model learns:  
$$\text{More study hours} \longrightarrow \text{Higher marks}$$

* **Output will be approximately:** `Predicted Marks: 85.0`

> 📌 **Remember:**  
> **Linear Regression** = Predict a number


````python
from sklearn.linear_model import LinearRegression

# Data
X = [[1], [2], [3], [4], [5]]  #Why [[1], [2], [3]] instead of [1, 2, 3]? Because Scikit-learn expects the input in 2D format: Rows = students/data records Columns = features
y = [35, 45, 55, 65, 75]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)  #fit() means: Learn from the given data.

# Predict marks for 6 study hours
prediction = model.predict([[6],[7]])

print("Predicted Marks:", prediction[0],prediction[1])
````

## 2. Logistic Regression

### What does it do?
**Logistic Regression is used for classification (predicting a category).**

* **Example:** Will the student pass or fail?

\[\text{Key:} \quad \mathbf{0 = \text{Fail}} \quad \text{and} \quad \mathbf{1 = \text{Pass}}\]

---

### The Data

| Study Hours | Result |
| :---: | :---: |
| 1 | 0 |
| 2 | 0 |
| 3 | 0 |
| 4 | 1 |
| 5 | 1 |
| 6 | 1 |

---

### The Program

```python
from sklearn.linear_model import LogisticRegression

# Data
X = [[1], [2], [3], [4], [5], [6]]
y = [0, 0, 0, 1, 1, 1]

# Create model
model = LogisticRegression()

# Train
model.fit(X, y)

# Predict for 5 study hours
prediction_5 = model.predict([[5]])
print("Prediction for 5 hours:", prediction_5[0]) 
# Output: 1 (Meaning: Pass)

# Predict for 2 study hours
prediction_2 = model.predict([[2]])
print("Prediction for 2 hours:", prediction_2[0]) 
# Output: 0 (Meaning: Fail)
```

---

### Simple Explanation
$$\text{Study Hours} \longrightarrow \text{Logistic Regression} \longrightarrow \text{Category (0 or 1)}$$

The model sets a boundary line to sort students into two buckets based on their hours. 

* **5 Hours** $\longrightarrow$ Placed in the **1 (Pass)** bucket.
* **2 Hours** $\longrightarrow$ Placed in the **0 (Fail)** bucket.

> 📌 **Remember:**  
> **Logistic Regression** = Classification (Predict a category)


````python
from sklearn.linear_model import LogisticRegression

# Data
X = [[1], [2], [3], [4], [5], [6]]
y = [0, 0, 0, 1, 1, 1]

# Create model
model = LogisticRegression()

# Train
model.fit(X, y)

# Predict for 5 study hours
prediction_5 = model.predict([[7]])
print("Prediction for 5 hours:", prediction_5[0]) 
# Output: 1 (Meaning: Pass)

# Predict for 2 study hours
prediction_2 = model.predict([[9]])
print("Prediction for 2 hours:", prediction_2[0]) 
# Output: 0 (Meaning: Fail)
````

## 3. Decision Tree

### What does it do?
**A Decision Tree makes decisions by asking a series of True/False questions.**

* **Example concept:**
```text
      [ Study Hours > 3? ]
               │
       ┌───────┴───────┐
       ▼               ▼
      No              Yes
    (Fail)          (Pass)
```

---

### The Program

```python
from sklearn.tree import DecisionTreeClassifier

# Input: Age
X = [[18], [20], [22], [25], [30], [35]]

# Output: No = 0, Yes = 1
y = [0, 0, 1, 1, 1, 1]

# Create Decision Tree
model = DecisionTreeClassifier()

# Train the model
model.fit(X, y)

# Predict for a new person
prediction = model.predict([[25]])

print("Prediction:", prediction[0])
```

---

### Visualize the Tree Graphically
*(Run this in a standard Python cell right below your code to see the actual visual tree structure)*

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))

plot_tree(
    model,
    feature_names=["Study Hours"],
    class_names=["Fail", "Pass"],
    filled=True
)

plt.show()
```

---

> 📌 **Remember:**  
> **Decision Tree** = Asking questions to split data and make a final decision.


````python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier

# Data
X = [[1], [2], [3], [4], [5], [6]]
y = [0, 0, 0, 1, 1, 1]

# Create model
model = DecisionTreeClassifier(random_state=42)

# Train
model.fit(X, y)

# Predict for 5 study hours
prediction = model.predict([[5]])

print("Prediction:", prediction[0])

plt.figure(figsize=(6, 4))

plot_tree(
    model,
    feature_names=["Study Hours"],
    class_names=["Fail", "Pass"],
    filled=True
)

plt.show()
````

## 1. What is OOP?

**OOP (Object-Oriented Programming)** is a way of writing programs using **classes and objects**.

A simple real-world example:

- **Class** → Blueprint of a car
- **Object** → Actual car created from that blueprint
- **Attributes** → Color, brand, model
- **Methods** → Start, stop, drive


## 2. Class and Object

A **class** is a blueprint. An **object** is an instance of a class.

```python
class Student:
    pass

# Creating an object
student1 = Student()

print(student1)
```

### Easy explanation

- `class Student:` creates a class.
- `student1 = Student()` creates an object.
- `student1` is an object of the `Student` class.

## 3. Constructor `__init__()`

The `__init__()` method runs automatically when we create an object.

It is commonly used to give values to an object's attributes.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Pravin", 24)

print("Name:", student1.name)
print("Age:", student1.age)
```

### What is `self`?

`self` refers to the **current object**.

For example:

`self.name = name`

means the `name` value belongs to the current object.

## 4. Attributes and Methods

- **Attribute** → Data or property of an object
- **Method** → Function inside a class

```python
class Student:
    def __init__(self, name, age):
        self.name = name      # Attribute
        self.age = age        # Attribute

    def introduce(self):      # Method
        print("My name is", self.name)
        print("I am", self.age, "years old")


student1 = Student("Pravin", 24)
student1.introduce()
```

## 5. Creating Multiple Objects

One class can be used to create many objects.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(self.name, "is", self.age, "years old")


student1 = Student("Pravin", 24)
student2 = Student("Rahul", 23)

student1.introduce()
student2.introduce()
```

## 6. Encapsulation

**Encapsulation** means keeping data and the methods that work with that data together inside a class.

A simple example uses a private variable with `__`.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def show_balance(self):
        print("Balance:", self.__balance)

    def deposit(self, amount):
        self.__balance += amount


account = BankAccount(1000)
account.show_balance()

account.deposit(500)
account.show_balance()
```

### Easy explanation

`__balance` is intended to be private to the class.

Instead of changing the balance directly, we use methods such as `deposit()`.

## 7. Inheritance

**Inheritance** allows one class to use the properties and methods of another class.

Here, `Dog` inherits from `Animal`.

```python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


dog = Dog()
dog.eat()    # Inherited method
dog.bark()   # Dog's own method
```

### Easy explanation

`Dog(Animal)` means the `Dog` class inherits from `Animal`.

Therefore, a `Dog` object can use the `eat()` method from `Animal`.

## 8. Polymorphism

**Polymorphism** means the same method name can behave differently for different objects.

```python
class Dog:
    def sound(self):
        print("Dog says: Woof")


class Cat:
    def sound(self):
        print("Cat says: Meow")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

### Easy explanation

Both classes have a method called `sound()`.

- Dog → `Woof`
- Cat → `Meow`

Same method name, different behavior.

## 9. Abstraction

**Abstraction** means hiding unnecessary implementation details and showing only what is needed.

Python provides the `abc` module for creating abstract classes.

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):
    def start(self):
        print("Car starts with a key")


car = Car()
car.start()
```

### Easy explanation

`Vehicle` says that every vehicle must have a `start()` method.

The `Car` class provides the actual implementation.

# 10. Complete Simple Example

Let's combine class, object, constructor, attributes, and methods in one small program.

```python
class Laptop:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def show_details(self):
        print("Brand:", self.brand)
        print("Price:", self.price)


laptop1 = Laptop("Dell", 50000)
laptop2 = Laptop("HP", 45000)

laptop1.show_details()
print()
laptop2.show_details()
```

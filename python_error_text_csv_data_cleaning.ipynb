# Part 1 — Error Handling

## 2. What is an Error?

An error happens when Python cannot perform an operation.

Example: dividing a number by zero.

````python
# This code produces an error if we run it directly.
# 10 / 0
````

## 3. `try` and `except`

We put risky code inside `try`.

If an error happens, `except` handles it.

````python
try:
    number = 10 / 0
except ZeroDivisionError:
    print("You cannot divide by zero!")
````

### Easy idea

Think of it like this:

```text
TRY this work
   ↓
Did an error happen?
   ↓
YES → EXCEPT handles it
NO  → continue normally
```

## 4. Handling Different Errors

Here we handle a value that cannot be converted to an integer.

````python
try:
    age = int("hello")
except ValueError:
    print("Please enter a valid number.")
````

## 5. `else` and `finally`

- `else` runs when there is **no error**.
- `finally` runs **every time**, whether there is an error or not.

````python
try:
    number = int("25")
except ValueError:
    print("Invalid number")
else:
    print("Conversion successful:", number)
finally:
    print("This block always runs.")
````

# Part 2 — Reading and Writing Text Files

## 6. Writing a Text File

We can use `open()` with mode `"w"`.

`w` means **write**. If the file already exists, its old content will be replaced.

````python
with open("students.txt", "w", encoding="utf-8") as file:
    file.write("Arun - 85\n")
    file.write("Priya - 92\n")
    file.write("Kumar - 78\n")

print("File created successfully!")
````

## 7. Reading a Text File

Use mode `"r"` to read a file.

````python
with open("students.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

## 8. Read a File Line by Line

This is useful when a file contains many lines.

````python
with open("students.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())
````

## 9. Append New Data

Use mode `"a"` to add content without deleting the existing content.

````python
with open("students.txt", "a", encoding="utf-8") as file:
    file.write("Divya - 88\n")

print("New student added!")
````

## 10. Safely Read a File with Error Handling

If the file does not exist, Python raises `FileNotFoundError`.

````python
try:
    with open("unknown_file.txt", "r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found. Please check the file name.")
````

# Part 3 — Working with CSV Data

## 11. What is a CSV File?

**CSV = Comma-Separated Values**.

Example:

```text
Name,Age,City,Salary
Arun,22,Chennai,25000
Priya,25,Chennai,30000
Kumar,28,Coimbatore,35000
```

CSV files are commonly used to store datasets.

## 12. Create a Small CSV Dataset

We will create a messy dataset so we can practice data cleaning.

````python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Arun", "Kumar", "Divya", "Ravi"],
    "Age": [22, 25, 22, None, 28, -5],
    "City": ["Chennai", "chennai", "Coimbatore", "Coimbatore", None, "Madurai"],
    "Salary": [25000, 30000, 25000, 35000, None, 40000]
}

df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)

print("CSV file created!")
````

## 13. Read a CSV File

Pandas makes CSV reading very easy.

````python
df = pd.read_csv("students.csv")
df
````

## 14. Understand the CSV Data

````python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
````

# Part 4 — Data Cleaning

## 15. Find Missing Values

`isnull().sum()` tells us how many missing values each column has.

````python
df.isnull().sum()
````

## 16. Fill Missing Age

We can use the median age to fill the missing value.

````python
age_median = df["Age"].median()
df["Age"] = df["Age"].fillna(age_median)

print("Median age:", age_median)
df
````

## 17. Fill Missing City

For text data, we can use the most common value (mode).

````python
city_mode = df["City"].mode()[0]
df["City"] = df["City"].fillna(city_mode)

print("Most common city:", city_mode)
df
````

## 18. Fill Missing Salary

Salary is numerical, so median is a simple choice.

````python
salary_median = df["Salary"].median()
df["Salary"] = df["Salary"].fillna(salary_median)

print("Median salary:", salary_median)
df
````

## 19. Remove Duplicate Rows

Arun appears twice with the same data. We can remove duplicate rows.

````python
print("Number of duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("After removing duplicates:", df.duplicated().sum())
df
````

## 20. Standardize Text Data

`Chennai` and `chennai` should be treated as the same city.

We can use `.str.title()`.

````python
df["City"] = df["City"].str.strip().str.title()
df
````

## 21. Find Incorrect Values

Ravi has an age of `-5`.

A negative age is not valid, so we need to fix it.

````python
print("Invalid ages:")
print(df[df["Age"] < 0])
````

## 22. Replace the Incorrect Age

We replace the negative age with the median of valid ages.

````python
valid_age_median = df.loc[df["Age"] >= 0, "Age"].median()
df.loc[df["Age"] < 0, "Age"] = valid_age_median

print("Replacement age:", valid_age_median)
df
````

## 23. Clean Column Names

Good column names make our Python code easier to read.

````python
df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_")
print(df.columns.tolist())
````

## 24. Final Data Quality Check

After cleaning, always check the data again.

````python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nInvalid ages:", (df["age"] < 0).sum())

print("\nClean dataset:")
display(df)
````

## 25. Save the Clean CSV

Use `to_csv()` to save the cleaned dataset.

````python
df.to_csv("students_clean.csv", index=False)
print("Clean CSV saved successfully!")
````
